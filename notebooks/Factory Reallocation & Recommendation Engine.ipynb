```python
from math import radians, sin, cos, sqrt, atan2

def haversine(lat1, lon1, lat2, lon2):

    R = 6371

    lat1, lon1, lat2, lon2 = map(
        radians,
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2-lat1
    dlon = lon2-lon1

    a = (
        sin(dlat/2)**2 +
        cos(lat1)*cos(lat2)*sin(dlon/2)**2
    )

    c = 2*atan2(
        sqrt(a),
        sqrt(1-a)
    )

    return R*c
```

```python
features = [
    'Ship Mode',
    'Division',
    'Region',
    'Product Name',
    'Factory',
    'Shipping Distance (km)',
    'Sales',
    'Units',
    'Cost',
    'Gross Profit',
    'Profit Margin'
]
```

```python
def predict_lead_time(new_row):

    # Create a one-row DataFrame
    sample_df = pd.DataFrame([new_row])

    # Keep only training features
    sample_df = sample_df[features]

    # Encode categorical columns
    categorical_columns = [
        'Ship Mode',
        'Division',
        'Region',
        'Product Name',
        'Factory'
    ]

    for col in categorical_columns:
        sample_df[col] = label_encoders[col].transform(
            sample_df[col].astype(str)
        )

    # Scale numerical columns
    numeric_columns = [
        'Shipping Distance (km)',
        'Sales',
        'Units',
        'Cost',
        'Gross Profit',
        'Profit Margin'
    ]

    sample_df[numeric_columns] = scaler.transform(
        sample_df[numeric_columns]
    )

    # Predict lead time
    prediction = model.predict(sample_df)[0]

    return prediction
```

```python
predicted_lead_time = predict_lead_time(new_row)

print("Predicted Lead Time:", round(predicted_lead_time, 2), "days")
```

```python
recommendations = []

for product in products:

    product_data = df[df["Product Name"] == product].copy()

    sample = product_data.iloc[0]

    current_factory = sample["Factory"]

    current_prediction = predict_lead_time(sample)

    for factory in factory_list:

        new_row = sample.copy()

        new_row["Factory"] = factory

        new_row["Factory Latitude"] = factory_coordinates[factory][0]
        new_row["Factory Longitude"] = factory_coordinates[factory][1]

        distance = haversine(
            new_row["Factory Latitude"],
            new_row["Factory Longitude"],
            new_row["Customer Latitude"],
            new_row["Customer Longitude"]
        )

        new_row["Shipping Distance (km)"] = distance

        predicted = predict_lead_time(new_row)

        recommendations.append({
            "Product": product,
            "Current Factory": current_factory,
            "Recommended Factory": factory,
            "Current Lead Time": current_prediction,
            "Predicted Lead Time": predicted,
            "Shipping Distance (km)": distance,
            "Lead Time Improvement": current_prediction - predicted,
            "Profit Margin": sample["Profit Margin"]
        })
```

```python
recommendation_df = pd.DataFrame(recommendations)

recommendation_df.head()
```

```python
recommendation_df["Optimization Score"] = (
    recommendation_df["Lead Time Improvement"] * 0.60
    - recommendation_df["Shipping Distance (km)"] * 0.0002
    + recommendation_df["Profit Margin"] * 0.15
)
```

```python
recommendation_df = recommendation_df.sort_values(
    ["Product", "Optimization Score"],
    ascending=[True, False]
)
```

```python
best_recommendations = (
    recommendation_df
    .groupby("Product")
    .head(1)
    .reset_index(drop=True)
)
```

```python
recommendations = []

for product in products:

    product_orders = df[df["Product Name"] == product]

    current_factory = product_factory_map[product]

    for factory in factory_list:

        lead_times = []
        distances = []
        profit_margins = []

        for _, row in product_orders.iterrows():

            new_row = row.copy()

            new_row["Factory"] = factory

            new_row["Factory Latitude"] = factory_coordinates[factory][0]
            new_row["Factory Longitude"] = factory_coordinates[factory][1]

            distance = haversine(
                new_row["Factory Latitude"],
                new_row["Factory Longitude"],
                new_row["Customer Latitude"],
                new_row["Customer Longitude"]
            )

            new_row["Shipping Distance (km)"] = distance

            predicted = predict_lead_time(new_row)

            lead_times.append(predicted)
            distances.append(distance)
            profit_margins.append(new_row["Profit Margin"])

        recommendations.append({

            "Product": product,

            "Current Factory": current_factory,

            "Factory": factory,

            "Average Lead Time": np.mean(lead_times),

            "Average Distance": np.mean(distances),

            "Average Profit Margin": np.mean(profit_margins)

        })
```

```python
recommendation_df = pd.DataFrame(recommendations)

recommendation_df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

score_scaler = MinMaxScaler()

recommendation_df[
    [
        "LeadTimeScore",
        "DistanceScore",
        "ProfitScore"
    ]
] = score_scaler.fit_transform(

    recommendation_df[
        [
            "Average Lead Time",
            "Average Distance",
            "Average Profit Margin"
        ]
    ]

)
```

```python
recommendation_df["Optimization Score"] = (

    (1 - recommendation_df["LeadTimeScore"]) * 0.50 +

    (1 - recommendation_df["DistanceScore"]) * 0.30 +

    recommendation_df["ProfitScore"] * 0.20

)
```

```python
recommendation_df = recommendation_df.sort_values(
    ["Product", "Optimization Score"],
    ascending=[True, False]
)
```

```python
best_recommendations = (

    recommendation_df
    .groupby("Product")
    .first()
    .reset_index()

)

best_recommendations
```
